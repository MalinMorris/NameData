# Predicting Rankings 2025
I will attempt to use the trends of the top 12 boy and girl names in 2024 to predict the 2025 rankings for the top 10 boy and girl names.

```python
import functions as fun
fun.init(1880, 2026, "names")
girls_2024 = ['Olivia', 'Emma', 'Amelia', 'Charlotte', 'Mia', 'Sophia', 'Isabella', 'Evelyn', 'Ava', 'Sofia',
              'Camila', 'Harper']
boys_2024 = ['Liam', 'Noah', 'Oliver', 'Theodore', 'James', 'Henry', 'Mateo', 'Elijah', 'Lucas', 'William',
             'Benjamin', 'Levi']
```

```python
color_index = 0
for n in girls_2024:
    name_counts, ignore = f.name_counts_years(n.lower(), 'f', 2000, 2025)
    plt.plot(range(2000,2025), name_counts, 
             color = colors[color_index], label = f"{n}, (f)")
    color_index += 1
plt.title("Top 12 Girl Names 2024")
plt.xlabel('Year')
plt.ylabel('Count')
plt.grid()
plt.legend()
```

```python
color_index = 0
for n in boys_2024:
    name_counts, ignore = f.name_counts_years(n.lower(), 'm', 2000, 2025)
    plt.plot(range(2000,2025), name_counts, 
             color = colors[color_index], label = f"{n}, (m)")
    color_index += 1
plt.title("Top 12 Boy Names 2024")
plt.xlabel('Year')
plt.ylabel('Count')
plt.grid()
plt.legend()
```
