```python
def calculate_spread(name):
    """Takes in name, finds counts by year for each sex
    returns percent difference, number of years with any data, male and female counts,
    and a metric of when the data is centered"""
    # do name counts years
    f_counts, _ = fun.name_counts_years(name, "f", min_y=min_data_year, max_y=max_data_year)
    m_counts, _ = fun.name_counts_years(name, "m", min_y=min_data_year, max_y=max_data_year)
    # do percent difference
    num_f = sum(f_counts)
    num_m = sum(m_counts)
    per_diff = fun.calculate_percent_diff(num_f, num_m)
    # add lists together
    counts = [f_counts[i] + m_counts[i] for i in range(len(f_counts))]
    # find center of data by finding year where sum surpasses current halfway point
    years = sum(1 for i in counts if i != 0)
    half_count = sum(counts) / 2
    s = 0
    center = -1
    for i in range(len(counts)):
        s += counts[i]
        if s >= half_count:
            center = i + min_data_year
            break
    return center, years, per_diff, num_f, num_m
```

## Extinct Names
Finding names that used to be used but no longer are at four levels:
- Vulnerable -- names in the top 1,000 from 1945 or earlier with 50 or less per year since 1960
- Endangered -- names in the top 1,000 from 1945 or earlier with 20 or less per year since 1960
- Critically Endangered -- names in the top 1,000 from 1945 or earlier with 10 or less per year since 1960
- Extinct -- names in the top 1,000 from 1945 or earlier with no uses since 1960

```python
f_top_1000_25 = set() # female names from top 1,000 from 1925 or earlier
m_top_1000_25 = set()
names_25 = {"f" : f_top_1000_25, "m" : m_top_1000_25}
f_top_1000_45 = set() # female names from top 1,000 from 1945 or earlier
m_top_1000_45 = set()
names_45 = {"f" : f_top_1000_45, "m" : m_top_1000_45}
ranks = ["extinct", "crit", "endangered", "vulnerable", "fine"]
```

```python
def names_sorter(sex, all_names, min_y=min_data_year, max_y=max_data_year):
    """Checks if the counts of a name across min_y to max_y are less than a number"""
    endangered = pd.DataFrame(columns = ["Name", "Sex", "Rating", "Count"])
    candidate_names = {x : 0 for x in all_names} # 0 = extinct
    for year in range(min_y, max_y): # iterate over years name must follow criteria
        year_list = fun.get_year_list_one(sex, year)
        for n in list(candidate_names): # iterate over names
            if n in list(year_list["Name"]):
                count = year_list[year_list["Name"] == n]["Count"].sum() 
            else:
                count = 0
            current_rank = candidate_names[n]
            if count < 5 and current_rank < 1:
                pass
            elif count <= 10 and current_rank < 2:
                candidate_names[n] = 1 # critically endangered
            elif count <= 20 and current_rank < 3:
                candidate_names[n] = 2 # endangered
            elif count <= 50 and current_rank < 4:
                candidate_names[n] = 3 # vulnerable
            else:
                candidate_names[n] = 4 # fine
    for n, r in candidate_names.items():
        if r != 4:
            total_count = fun.name_counts_simple(n, sex)[0]
            endangered.loc[len(endangered)] = [n, sex, ranks[r], total_count]
    return endangered
```
